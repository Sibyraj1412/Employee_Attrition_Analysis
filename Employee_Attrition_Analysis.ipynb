## Attrition Distribution
This chart shows the number of employees who stayed and left the company.

```python
sns.countplot(data=df, x="Attrition")
plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

## Attrition by Department
This chart helps us identify which departments have more employee attrition.

```python
sns.countplot(data=df, x="Department", hue="Attrition")
plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=15)
plt.show()
```

## Age Distribution
This histogram shows the age distribution of employees.

```python
sns.histplot(data=df, x="Age", hue="Attrition", bins=20)
plt.title("Age Distribution by Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

## Monthly Income by Attrition
This box plot compares the monthly income of employees who stayed and left.

```python
sns.boxplot(data=df, x="Attrition", y="MonthlyIncome")
plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")
plt.show()
```

## Overtime and Attrition
This chart helps us understand the relationship between overtime and employee attrition.

```python
sns.countplot(data=df, x="OverTime", hue="Attrition")
plt.title("Employee Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")
plt.show()
```

```python
# Insight 1: Overall Attrition
attrition_count = df["Attrition"].value_counts()
print("1. Employee Attrition:")
print(attrition_count)

# Insight 2: Attrition by Department
department_attrition = pd.crosstab(
    df["Department"], df["Attrition"]
)
print("\n2. Attrition by Department:")
print(department_attrition)

# Insight 3: Attrition by Overtime
overtime_attrition = pd.crosstab(
    df["OverTime"], df["Attrition"]
)
print("\n3. Attrition by Overtime:")
print(overtime_attrition)

# Insight 4: Attrition by Age Group
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)

age_attrition = pd.crosstab(
    df["AgeGroup"], df["Attrition"]
)
print("\n4. Attrition by Age Group:")
print(age_attrition)

# Insight 5: Attrition by Monthly Income
income_attrition = df.groupby("Attrition")["MonthlyIncome"].mean()
print("\n5. Average Monthly Income by Attrition:")
print(income_attrition)
```

# Finding Five Insights

### How many employees left the company?

```python
# Number of employees who left
print("Employees who left:")
print(df["Attrition"].value_counts())
```

### Which department has more employees leaving?

```python
# Attrition by department
print("\nAttrition by Department:")
print(pd.crosstab(df["Department"], df["Attrition"]))
```

### Are employees who work overtime more likely to leave?

```python
# Attrition by overtime
print("\nAttrition by Overtime:")
print(pd.crosstab(df["OverTime"], df["Attrition"]))
```

### Which age group has more employee attrition?

```python
# Attrition by age group
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"],
    include_lowest=True
)
print("\nAttrition by Age Group:")
print(pd.crosstab(df["AgeGroup"], df["Attrition"]))
```

### Do employees who leave have lower average salaries?

```python
# Average monthly income
print("\nAverage Monthly Income:")
print(df.groupby("Attrition")["MonthlyIncome"].mean())
```
